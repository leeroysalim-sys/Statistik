Dataset source: 'https://raw.githubusercontent.com/CarrenP/files/refs/heads/main/penguins_size.csv'

```python
# import the df from the link above
url = 'https://raw.githubusercontent.com/CarrenP/files/refs/heads/main/penguins_size.csv'
df = pd.read_csv(url, sep = ",")
df.head()
```

```python
# menunjukan informasi mengenai tabel, dari data type, jumlah data hingga nama kolum.
df.info()
```

```python
# Count the penguins based on their species
df['species'].value_counts()
```

```python
# create a pie chart to visualise the species, explain your finding in at least 2 sentences.
df['island'].value_counts().plot(
kind='pie', figsize=(10,6),
autopct='%1.2f%%', legend=True
)
plt.title('Penguin Distribution by Island')
plt.show()

# Explanation:
# The pie chart shows the proportion of each island in the dataset. 
# Biscoe appears to be the most common island, followed by Torgersen and Dream, 
# indicating a varied distribution across the islands.
```

```python
# Display the data distribution of culmen length by island using box plot,explain your findng in at least 2 setences.
df.boxplot(by='island', column=['culmen_length_mm'], grid=False)
plt.show()

# Explanation:
# The box plot illustrates the distribution of culmen length across different islands.
# It appears that Biscoe has a wider range of culmen lengths compared to Torgersen and Dream,
# which may suggest more variability in the penguin population on Biscoe.
```
