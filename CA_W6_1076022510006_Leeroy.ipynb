```python
# Import necessary library

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

### Frequency Table
Grouping of qualitative data into mutually exclusive classes showing the number of observations in each class.<br>
Dataset source: 'https://raw.githubusercontent.com/twiradinata/datasets/main/applewood_auto_group.csv'

```python
url = 'https://raw.githubusercontent.com/twiradinata/datasets/main/applewood_auto_group.csv'
df = pd.read_csv(url, sep = ";")
df.head()
```

```python
# Display the data types and structure
df.info()
```

```python
# Creating frequency table of total purchases using groupby function
df_total = df.groupby('Location')['Purchases'].sum()
df_total 
```

```python
# Creating frequency table of total count using value_counts function, which one would you use?
# The concise, "best" way for this specific output:
df['Location'].value_counts()
```

### Bar Chart to Visualize Discrete Data

```python
# Create a bar plot to visualize number of purchase by location

ax = df_total.plot(kind="bar")
ax.bar_label(ax.containers [0])
plt.title("Purchase by Location", pad=20)
plt.xlabel("Location")
plt.ylabel("Purchases")
plt.show()
```

### Stem and Leaf Plot to Visualize Frequency Distribution
Stem-and-leaf plot is a tabular presentation where each data value is split into a “stem” (the first digit or digits) and a “leaf” (usually the last digit).

```python
# Visualize a frequency distribution using stem-and-leaf plot
from stemgraphic import stem_graphic
stem_graphic (df ['Age of buyer'], scale = 10)
plt.title('Stem and Leaf Plot for Age of Buyer', pad=20)
plt.show()
```

### Scatter (Dot) Plot to Visualize Relationship

```python
# Create a scatter plot to visualize the relationship between profit and age of buyer
df_profit = df. loc [df ['Profit'] < 200]
df_profit.plot.scatter (x = 'Age of buyer', y = 'Profit')
plt.title('Relationship between Profit and Age of Buyer', pad=20)
plt.show()
```

### Pie Chart to Visualize Composition

```python
# Make a visualization of purchase composition by location
df.groupby('Location') ['Purchases'].sum().plot(
kind='pie', y='Purchases', figsize=(10,6),
autopct='%1.2f%%', legend=True
)
plt.title('Purchase Composition among Location')
plt.show()
```

### Using Box Plot to visualize data distribution

```python
# Display the data distribution of Age of Buyer by Location using box plot
df.boxplot(by='Location', column=['Age of buyer'], grid=False)
plt.show()
```

### Using Histogram to visualize data distribution

```python
# Pay attention to the Age of Buyer data, histogram shows a normal distribution
df ['Age of buyer'].plot(kind='hist')
plt.show()
```

### Outlier Detection and Removal

```python
# Find out from the internet how to remove outlier and redisplay the box plot
Q1 = df ['Age of buyer'].quantile(0.25)
Q3 = df ['Age of buyer'].quantile(0.75) 
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df_no_outliers = df [(df ['Age of buyer'] >= lower_bound) & (df ['Age of buyer'] <= upper_bound)]
df_no_outliers.boxplot(by='Location', column=['Age of buyer'], grid=False)
plt.title('Box Plot of Age of Buyer by Location (Without Outliers)')
plt.show()
```

### Measures of Position

### Utilizing Numpy's percentile function to find data position

```python
# Find the Percentiles of commissions earned

commissions = [
    2038, 1758, 1721, 1637, 
    2097, 2047, 2205, 1787, 
    2287, 1940, 2311, 2054, 
    2406, 1471, 1460
]

median = np.percentile(commissions, 50)  
q1 = np.percentile(commissions, 25)    
q3 = np.percentile(commissions, 75)      

print(f"50th percentile of commissions : {median}")
print(f"25th percentile of commissions : {q1}")
print(f"75th percentile of commissions : {q3}")
```

### Box Plot to Visualize data distribution in a variable

```python
# Visualize the percentiles of commissions earned using box plot

plt.figure(figsize=(10, 6))
plt.boxplot(commissions, vert=False, patch_artist=True)

plt.title('Visualizing Percentiles of Commissions Earned')
plt.xlabel('Commission Amount ($)')
plt.yticks([1], ['0'])
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.show()
```

Create several visualizations to describe the condition of the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

url = 'https://raw.githubusercontent.com/twiradinata/datasets/main/mlbb_hero.csv'
df = pd.read_csv(url, sep=";")
df.head()
```

```python
# Display the data types and structure
df['hero_name'] = df['hero_name'].astype('object')
df['role'] = df['role'].astype('object')
df.info()
```

```python
ax = df['role'].value_counts().plot(kind="bar")
ax.bar_label(ax.containers [0])
plt.title("Number of Heroes by Role", pad=20)
plt.xlabel("Role")
plt.ylabel("Number of Heroes")
plt.show()
```

```python
# Observe data distribution of movement_spd for each role using box plot
df.boxplot(by='role', column=['movement_spd'], grid=False)
plt.title('Box Plot of Movement Speed by Role', pad=20)
plt.xlabel('Role')
plt.ylabel('Movement Speed')
plt.show()
```

```python
# Observe, is there relationship between physical_atk and win_rate
df.plot.scatter(x='physical_atk', y='win_rate')
plt.title('Relationship between Physical Attack and Win Rate', pad=20)
plt.xlabel('Physical Attack')
plt.ylabel('Win Rate (%)')
plt.show()
```
