```python
# a). Calculate the average number of hours studied by the students.
hours_studied = [2, 4, 6, 8, 10]
sum_hours = sum(hours_studied)
num_students = len(hours_studied)
average_hours = sum_hours / num_students
print(f"The average number of hours studied by the students is: {average_hours} hours.")
```

```python
# b). Determine the weighted mean of the hours studied, considering the number of students in each group.
hours_studied = [2, 4, 6, 8, 10]
num_students = [3, 5, 2, 4, 1]  
weighted_sum = hours_studied[0] * num_students[0] + hours_studied[1] * num_students[1] + hours_studied[2] * num_students[2] + hours_studied[3] * num_students[3] + hours_studied[4] * num_students[4]
total_students = sum(num_students)
weighted_mean = weighted_sum / total_students
print(f"the weighted mean is: {weighted_mean}")
```

```python
# c). Calculate the range of the hours studied.
range_hours = max(hours_studied) - min(hours_studied)
print(range_hours)
```

```python
# d). Find the standard deviation of the hours studied to measure how spread out the data is.
hours_studied = [2, 4, 6, 8, 10]
n = len(hours_studied)
total = 0
for h in hours_studied:
    total += h
mean = total / n
sum_squared_diff = 0
for h in hours_studied:
    diff = h - mean
    sum_squared_diff += diff * diff
    variance = sum_squared_diff / n
std_dev = variance ** 0.5
print(std_dev)
```

```python
import matplotlib.pyplot as plt
import pandas as pd

data=pd.read_csv('https://raw.githubusercontent.com/CarrenP/files/main/NY-House-Dataset.csv')
data.head()
```

```python
#Create a bar chart to illustrate the Distribution of Houses by Sublocality in New York. explain your finding.
bar_chart = data['SUBLOCALITY'].value_counts().plot(kind='bar')
plt.title('Distribution of Houses by Sublocality in New York')
plt.xlabel('Sublocality')
plt.ylabel('Number of Houses')
plt.show()
```

```python
# Calculate the range of house prices in the dataset.
range_price = data['PRICE'].max() - data['PRICE'].min()
print("Maximum price: $", data['PRICE'].max())
print("Minimum price: $", data['PRICE'].min())
print("Range of prices: $", range_price)
```

```python
# Find the standard deviation of the house prices to understand the variability in pricing.
std_dev_price = data['PRICE'].std()
print("Standard Deviation of house prices: $", std_dev_price)
```

```python
# Calculate the average market price of the houses in the dataset.
avg=data['PRICE'].mean()
print("Average market price of the houses: $", avg)
```

```python
# Determine the mode for the number of bedrooms in the dataset.
mode=data['BEDS'].mode()
print("Mode for the number of bedrooms: ", mode[0])
```
