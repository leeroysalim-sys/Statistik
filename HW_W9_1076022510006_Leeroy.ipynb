## Classroom Exercise Week 8: Probability Distribution
A probability distribution is a mathematical function or model that describes the likelihood of different outcomes or events in a random or uncertain process. It specifies the probabilities of various possible outcomes, and the sum (or integral) of all these probabilities equals 1, indicating that one of the possible outcomes must occur.

There are two main types of probability distributions:
1. Discrete Probability Distribution: In this type of distribution, the random variable takes on distinct, separate values, often with gaps between them. Examples of discrete probability distributions include the binomial distribution, Poisson distribution, and the probability distribution of rolling a fair six-sided die.
2. Continuous Probability Distribution: Continuous distributions, on the other hand, deal with random variables that can take on any value within a specified range. The probabilities are associated with intervals or ranges of values rather than specific points. The normal distribution (also known as the Gaussian distribution) and the exponential distribution are examples of continuous probability distributions.

### Discrete Probability Distribution

John Ragsdale sells new cars for Pelican Ford. John usually sells the largest number of cars on Saturday. He has developed the following probability distribution for the number of cars he expects to sell on a particular Saturday. What is the mean and variance of the distribution?<br>


```python
example_of_carr_sold = [0, 1, 2, 3, 4]
prob_example_of_carr_sold = [0.1667, 0.2083, 0.4167, 0.1875, 0.0208]

example_data = {'example_of_carr_sold': example_of_carr_sold, 'prob_example_of_carr_sold': prob_example_of_carr_sold}
print(example_data) 
```

```python
df_example = pd.DataFrame(example_data)
df_example
```

```python
mean_for_example_data = 0

for i, j in zip(example_of_carr_sold, prob_example_of_carr_sold):
    mean_for_example_data = mean_for_example_data + (i * j)
    
print('The mean for example data is {:0.4f}'.format(mean_for_example_data))
```

```python
df_example['x-miu'] = df_example['example_of_carr_sold'] - mean_for_example_data
df_example
```

```python
df_example['(x-miu)^2'] = df_example['x-miu'] * df_example['x-miu']
df_example
```

```python
df_example['(x-miu)^2 * prob'] = df_example['(x-miu)^2'] * df_example['prob_example_of_carr_sold']
df_example
```

```python
variance = df_example['(x-miu)^2 * prob'].sum()
st_dev_example = np.sqrt(variance)

print('Variance: {:0.4f}'.format(variance))
print('Std Dev: {:0.4f}'.format(st_dev_example))
```

```python
num_of_cars_sold = [0, 1, 2, 3, 4]
prob_num_of_cars_sold = [0.1, 0.2, 0.3, 0.3, 0.1]
data = {
    'num_of_cars_sold': num_of_cars_sold,
    'prob_num_of_cars_sold': prob_num_of_cars_sold
}

df_num_of_cars_sold = pd.DataFrame(data)
df_num_of_cars_sold
```

```python
mean_of_prob_distribution = 0

for i, j in zip(num_of_cars_sold, prob_num_of_cars_sold):
    mean_of_prob_distribution += (i * j)

print('The mean of probability distribution is {:0.4f}'.format(mean_of_prob_distribution))
```

```python
df_num_of_cars_sold['(x-miu)'] = df_num_of_cars_sold['num_of_cars_sold'] - mean_of_prob_distribution
df_num_of_cars_sold['(x-miu)^2'] = np.power(df_num_of_cars_sold['(x-miu)'], 2)
df_num_of_cars_sold['(x-miu)^2 P(x)'] = df_num_of_cars_sold['(x-miu)^2'] * df_num_of_cars_sold['prob_num_of_cars_sold']
df_num_of_cars_sold
```

```python
variance_of_prob_distribution = df_num_of_cars_sold['(x-miu)^2 P(x)'].sum()
print('The variance of probability distribution is {:0.4f}'.format(variance_of_prob_distribution))
```

### Exercise 2
Belk Department Store is having a special sale this weekend. Customers charging purchases of more than $50 to their Belk credit card will be given a special Belk Lottery card. The customer will scratch off the card, which will indicate the amount to be taken off the total amount of the purchase. Listed below are the amount of the prize and the percent of the time that amount will be deducted from the total amount of the purchase. Compute the mean and standard deviation of the probability distribution.<br>


```python
prize_amount = [10, 25, 50, 100]
prob_prize_amount = [0.5, 0.4, 0.08, 0.02]
data_prize_amount = {
    'prize_amount': prize_amount,
    'prob_prize_amount': prob_prize_amount
}
df_prize_amount = pd.DataFrame(data_prize_amount)

mean_of_prob_distribution = 0
for i, j in zip(prize_amount, prob_prize_amount):
    mean_of_prob_distribution += (i * j)
print('The mean of probability distribution is {:0.4f}'.format(mean_of_prob_distribution))

df_prize_amount['(x-miu)'] = df_prize_amount['prize_amount'] - mean_of_prob_distribution
df_prize_amount['(x-miu)^2'] = np.power(df_prize_amount['(x-miu)'], 2)
df_prize_amount['(x-miu)^2 P(x)'] = df_prize_amount['(x-miu)^2'] * df_prize_amount['prob_prize_amount']

variance_of_prob_distribution = df_prize_amount['(x-miu)^2 P(x)'].sum()
print('The variance of probability distribution is {:0.4f}'.format(variance_of_prob_distribution)) 

st_dev_example = np.sqrt(variance_of_prob_distribution)
print('The standard deviation of probability distribution is {:0.4f}'.format(st_dev_example))
```
